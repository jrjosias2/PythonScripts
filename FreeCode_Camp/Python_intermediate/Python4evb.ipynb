```python
#Exercicio semana 6

def computepay(hours, rate):
    if hours >40:
        extra_hours = hours - 40
        extra_hour_rate = (extra_hours + (extra_hours/2)) * rate
        journey_payment = 40 * rate
        return journey_payment + extra_hour_rate
    else:
        return hours * rate

journey_hours = input("Enter Hours: ")
rating_card = input("Enter rate: ")
print(f'Pay {computepay(int(journey_hours), float(rating_card))}')
```

```python
dic_sender = dict()
bigcount = None
bigword = None

name = input("Enter file:")

if len(name) < 1:
    name = "mbox-short.txt"
else:
    name = name
    
handle = open(name)

'''
Linha que quebra as linhas e salva somente o endereço de email ao dicionario, adicionando um count a cada nova ocorrencia
No segundo loop de for temos uma variavel de controle para o endereco do email outra para a contagem de ocorrencia no arquivo
----
Quando inicia o loop a variavel da Maior ocorrencia e Nulo, entao entra no loop
Na segunda iteracao o bigcount tem registrado o valor do primeiro registro do dicionario, assim ele so vai entrar se o valor novo e maior que o anterior
Quando for maior que o anterior, atualiza o valor do contador e também o nome do contato de email a quem pertence esse valor.

----
Utilizando o Max não funcionou, porque utilizando o Max diretamente, ele pega o maior valor, entao pro Count dava certo, mas para o valor da Key, pega o de maior tamanho
e isso gerava erro na comparacao.
'''
for line in handle:
    if line.startswith("From "):
        dic_sender[line.split()[1]] = dic_sender.get(line.split()[1], 0) +1

print(dic_sender)

for word, count in dic_sender.items():
    if bigcount is None or count > bigcount:
        bigword = word
        bigcount = count

print(bigword, bigcount)
```
